Counting the frequencies in a list using a dictionary in Python

```python
lst = [1, 1, 1, 5, 5, 2, 3]
freq = {}
for item in lst:
    if item in freq:
        freq[item] += 1
    else:
        freq[item] = 1

print(freq)
```

Check whether a list contains a sublist

```python
def contains_sublist(list_1, list_2):
    return str(list_2)[1:-1] in str(list_1)
print(contains_sublist([1,2,3,4,5], [3,4]))
```

Replace 'not'...'poor' with 'good'

```python
str1 = input("enter string: ")
if "not" in str1 and "poor" in str1:
    start = str1.index("not")
    end = str1.index("poor") + 4
    str1 = str1.replace(str1[start:end], "good")
print(str1)
```

Count occurrences of each word in a given sentence

```python
sentence = input("Enter a sentence: ")
words = sentence.split()
word_count = {}
for word in words:
    word_count[word] = word_count.get(word, 0) + 1
print(word_count)
```

Sort dictionary by value

```python
d = {'a': 3, 'b': 1, 'c': 2}
asc = dict(sorted(d.items(), key=lambda x: x[1]))
desc = dict(sorted(d.items(), key=lambda x: x[1], reverse=True))
print(asc)
print(desc)
```

Find highest 3 values in a dictionary

```python
d = {'a': 10, 'b': 45, 'c': 12, 'd': 78, 'e': 5}
values = sorted(d.values(), reverse=True)
print(values[:3])
```

Given a number n, write a python program to make and print the list of Fibonacci
series up to n. Input : n=7 Hint : first 7 numbers in the series

```python
n = int(input("Enter n: "))
a, b = 0, 1
fib = []
for i in range(n + 1):
    fib.append(a)
    a, b = b, a + b
print(fib)
```

 Python Program to Find Factorial of Number Using Recursion
 

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n-1)
num = int(input("enter number: "))
print(factorial(num))
```
